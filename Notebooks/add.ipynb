```python
def add(a, b):
    c = a + b
    return c
```

```python
A = 2
B = 3
C = 5
print(add(A, B))
print(add(A, C))
print(add(B, C))
```

```python
def sum_values(vals):
    sum = 0
    i = 0
    while i < len(vals):
        sum = add(sum, vals[i])
        i += 1
    return sum        
```

```python
li = [1, 2, 103]
c = sum_values(li)
print(c)
```
